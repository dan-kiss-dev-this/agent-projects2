```python
# The imports

from dotenv import load_dotenv
from agents import Agent, Runner, trace
from agents.mcp import MCPServerStdio
from IPython.display import Image, display
import os
```

```python
tavily_params = {
    "command": "npx",
    "args": ["-y", "tavily-mcp@latest"],
    "env": {"TAVILY_API_KEY": os.getenv("TAVILY_API_KEY")},
}
```

```python

INSTRUCTIONS = "You are a job finding assistant, the job finding assistant helps reply to recruiters, research companies asked about and reply with concise high impact summaries and make emails for software engineer roles especially roles utilizing AI Agents"

TASK="Find out about https://www.alpinesg.com/ the company is called Alpine Software Group"

async with MCPServerStdio(params=tavily_params) as server_client:
    tavily_tools = await server_client.list_tools()
    for tool in tavily_tools:
        print(tool)
    print("--------------------------------")
   
    my_agent = Agent(
        "Researcher", 
        instructions=INSTRUCTIONS, 
        model="gpt-5.4-mini",
        mcp_servers=[server_client])
    with trace("investigate_ASG"):
        result = await Runner.run(my_agent, TASK, max_turns=5)
        print(result.final_output)
```
